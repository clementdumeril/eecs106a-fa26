### Building arrays by filling in entries

Instead of writing out every entry by hand, its sometimes easier to start from an array of all zeroes or the identity matrix, then fill in the only entries you care about.

```python
# Start with a 3x3 matrix of all zeros
rot = np.zeros((3, 3))
theta = np.pi / 4

# Fill in just the entries we need
rot[0, 0] = np.cos(theta)
rot[0, 1] = -np.sin(theta)
rot[1, 0] = np.sin(theta)
rot[1, 1] = np.cos(theta)
rot[2, 2] = 1

print(rot)

# np.eye(n) gives you an n x n identity matrix to start from
# instead of zeros, which is handy when most of the diagonal
# should stay 1
g = np.eye(4)
g[0, 3] = 5   # set the x-position
g[1, 3] = 2   # set the y-position
print(g)
```

### Slicing numpy arrays

Numpy uses pass-by-reference semantics so it creates views into the existing array, without implicit copying. This is particularly helpful with very large arrays because copying can be slow. Although be wary that you may be mutating an array when you don't intend to, so make sure to make a copy in these situations.

```python
orig = np.array([0, 1, 2, 3, 4, 5])
print(orig)
```

Slicing an array is just like slicing a list

```python
sliced = orig[1:4]
print(sliced)
```

Note, since slicing does not copy the array, mutating `sliced` mutates `orig`. Notice how the 4th element in `orig` changes to 9 as well.

```python
sliced[2] = 9
print(orig)
print(sliced)
```

We should use `np.copy()` to actually copy `orig` if we don't want to mutate it. 

```python
orig = np.array([0, 1, 2, 3, 4, 5])
copy = np.copy(orig)
sliced_copy = copy[1:4]
sliced_copy[2] = 9
print(orig)
print(sliced_copy)
```

### Slicing in more than one dimension

Everything above works in 1D arrays, but numpy slicing works in multiple dimensions: arr[row_slice, col_slice]. You can also assign directly into one of these slices to set a whole block of an array all at once. This comes in handy when building bigger matrices out of smaller matrices.

```python
g = np.eye(4)
rot = np.array([[0, -1, 0], [1, 0, 0], [0, 0, 1]])
position = np.array([1, 2, 3])

# Assign the top-left 3x3 block to be our matrix rot
g[:3, :3] = rot

# Assign the top 3 entries of the last column to be our array position
g[:3, 3] = position

print(g)

# You can read a block back out the same way
print(g[:3, :3]) 
print(g[:3, 3])
```

### Reshaping arrays

.reshape() lets you view the same data but with a different shape, which comes in handy when a function expects a specific shape for its input.

```python
v = np.array([1, 2, 3])
print(v.shape)          # (3,)

v_col = v.reshape(3, 1)
print(v_col.shape)      # (3, 1)
print(v_col)

# reshape doesn't copy data -- it's a view, just like slicing
v_col[0, 0] = 99
print(v)                # v is changed too!
```
